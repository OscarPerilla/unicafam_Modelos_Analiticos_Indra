```python
# ¿Te alcanza para el café?
presupuesto = 100000
precio_cafe = 2500
cafes_semana = 7
```

```python
# ¿Cuánto gastas al mes? (4 semanas)
gasto_mes = precio_cafe * cafes_semana * 4

print("Gastas al mes:", gasto_mes)
```

```python
# ¿Cuánto te sobra?
sobra = presupuesto - gasto_mes
print("Te sobra:", sobra)
```

```python
# ¿Te alcanza? (True o False)
alcanza = presupuesto >= gasto_mes
print("¿Te alcanza?:", alcanza)
```

```python
# Resumen general
print("Gastas al mes:", gasto_mes)
print("Te sobra:", sobra)
print("¿Te alcanza?:", alcanza)
```

¿ Te toca envío gratis?

Revisa tus ticket de compra

```python
# Precios de mi compra
precios = [250000, 382000, 50000, 632000, 95000]
```

```python
# Imprimir cada precio y el total
for precio in precios:
    print(precio)
```

```python
# Suma el total
for precio in precios:
    total = total + precio

print("Total:", total)
```

```python
# Contar cuantos cuestan mas de 100000
caros = 0
for precio in precios:
    if precio > 100000:
        caros = caros + 1

print("Productos mayores a 100000:", caros)
```
