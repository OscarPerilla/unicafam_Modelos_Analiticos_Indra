
**Fase 1**

* **Objetivo:** bajar los pedidos que llegan tarde en Bogotá para que los clientes queden más satisfechos, haya menos compensaciones y se repitan más compras.
* **KPI de éxito:** porcentaje de pedidos entregados a tiempo (por ejemplo, subir de 78 % a 90 %) y promedio de minutos de retraso frente al tiempo prometido.

**Fase 2**

* Horas de creación del pedido, aceptación del restaurante/comercio, salida del domiciliario y entrega real.
* Punto de recogida y punto de entrega (localidad/zona o coordenadas) y distancia calculada.
* Estado del tráfico y si está lloviendo en ese momento en Bogotá.
* Cantidad de domiciliarios disponibles en la zona y qué tan ocupado está el comercio en esa hora.
* Tipo/tamaño del pedido y medio de transporte del domiciliario.

**Fase 3**

* **Datos incompletos o marcas de tiempo mal registradas:** pedidos sin hora de recogida, domiciliarios que marcan entrega tarde o cuando ya se fueron, o fallas de señal GPS.
* **Duplicados o datos atípicos:** registros repetidos por reintentos en la app, tiempos imposibles (cero o negativos) o direcciones/zonas escritas de distintas formas.

**Fases 4–6**

* **Qué predecir:** si un pedido va a llegar tarde o cuántos minutos se demorará en total.
* **Cómo lo usaría la empresa:** mostrar tiempos estimados más realistas al pedir, asignar o mover domiciliarios antes de que se formen cuellos de botella (como en horas pico o lluvia) y dar alertas tempranas cuando un pedido vaya retrasado; primero se probaría en una zona o franja horaria para revisar si mejora el KPI antes de aplicarlo en todo el servicio.

```python
# Ejercicio 3: Primer análisis de ventas
```

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 200
```

```python
ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Bucaramanga", "pasto", "Armenia"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([35000, 46000, 89000], n),
})
```

```python
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
ventas.head()
```

```python
# ¿Cuál es el total vendido por canal (Tienda, Web, App)?

ventas.groupby("canal")["total"].sum()
```

```python
# ¿Cuál es el ticket promedio por ciudad?

ventas.groupby("ciudad")["total"].mean()
```

```python
# Días con mas ventas

ventas["dia_semana"] = ventas["fecha"].dt.day_name()

ventas.groupby("dia_semana")["total"].sum().idxmax()
```

El canal Web lidera las ventas, Bogotá registra la mayor facturación y Medellín presenta el ticket promedio más alto.
