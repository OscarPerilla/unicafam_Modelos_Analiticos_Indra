```python
# Importar pandas y numpy para manipulación de datos y cálculos numéricos (# indica comentario)
import pandas as pd
import numpy as np
```

```python
# Fijar la semilla aleatoria (42) para reproducibilidad y definir 200 registros
np.random.seed(42)
n = 200
```

```python
# Crear el DataFrame de ventas simuladas con fechas, ciudades, canales, unidades y precios
ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
```

```python
# Calcular el total de cada venta multiplicando unidades por precio
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
# Consultar dimensiones (filas, columnas) y mostrar las primeras 3 filas del DataFrame
print(ventas.shape)
print(ventas.head(3))
```

```python
# Revisar estructura, tipos de datos, valores no nulos y memoria utilizada
ventas.info()
```

Estadística descriptiva: ¿Qué pasó?

```python
# Agrupar ventas por ciudad y calcular conteo, suma y promedio, ordenando de mayor a menor suma
resumen = (ventas.groupby("ciudad")["total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))
```

```python
# Mostrar la tabla resumen por ciudad
resumen
```

```python
# Expresar el total en millones (1 decimal), redondear el promedio y renombrar columnas
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["ventas", "total_millones", "ticket_prom"]
```

```python
# Imprimir la tabla resumen con las columnas actualizadas
print(resumen)
```

```python
# Calcular métricas descriptivas básicas (media, desviación estándar, mínimos, cuartiles y máximos)
ventas.describe()
```

```python
# Construir tabla dinámica con la suma de ventas (en millones) por ciudad y canal
tabla = pd.pivot_table(ventas, values="total", index="ciudad",
                       columns="canal", aggfunc="sum") / 1e6
```

```python
# Mostrar la tabla redondeada a 1 decimal y obtener el canal con mayor total de ventas
print(tabla.round(1))
print()
print("Canal con más ventas:", ventas.groupby("canal")["total"].sum().idxmax())
```
